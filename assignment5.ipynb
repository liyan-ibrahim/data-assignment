```python
#filter out to see only if monsters are real 
monsters_df = scoobydoo_df.dropna(subset=['monster_real'])

monsters_df
```

```python
monsters_df['date_aired']
```

```python
# convert date strings into dates
monsters_df = monsters_df.copy()

monsters_df['year'] = pd.to_datetime(
    monsters_df['date_aired']
).dt.year
monsters_df[['title', 'year']]
```

```python
# decades to analyze trends within this incrememnt // potentially makes it less accurate 
monsters_df['decade'] = (monsters_df['year'] // 10) * 10

monsters_df[['title', 'year', 'decade']]
```

```python
# count real and fake monster classifications per decade
monster_counts = pd.crosstab(
    monsters_df['decade'],
    monsters_df['monster_real']
)

monster_counts
```

```python
# calculate percentage of real and fake monsters per decade
monster_percent = monster_counts.div(
    monster_counts.sum(axis=1),
    axis=0
) * 100

monster_percent
```

```python
#  real vs fake monsters over time in a visual cool way
monster_percent.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6),
    color=['#F5B5CC', '#A8CBB7']
)

plt.title('Real vs. Fake Monsters in Scooby-Doo Over Time')
plt.xlabel('Decade')
plt.ylabel('Percentage of Monsters (%)')
plt.legend(['Fake Monsters', 'Real Monsters'])
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
highest_decade = monster_percent[True].idxmax()

print("Decade with the highest percentage of real monsters:", highest_decade)

print("Percentage:", round(monster_percent[True].max(), 1), "%")
```
